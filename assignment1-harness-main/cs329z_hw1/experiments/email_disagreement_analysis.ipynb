```python
import json

TARGET_IDS = {
    "em-04260",
    "em-08472",
    "em-04298",
    "em-08173",
    "em-17561"
}

FILE_PATH = "email.jsonl"

found = set()

with open(FILE_PATH, "r", encoding="utf-8") as f:
    for line in f:
        if not line.strip():
            continue

        email = json.loads(line)

        if email.get("id") in TARGET_IDS:
            found.add(email["id"])

            print("\n" + "=" * 80)
            print("ID:", email.get("id"))
            print("From:", email.get("from"))
            print("To:", email.get("to"))
            print("Date:", email.get("date"))
            print("Subject:", email.get("subject"))
            print("Body:", email.get("body"))

print("\nFound:", len(found), "out of", len(TARGET_IDS))
print("Missing IDs:", TARGET_IDS - found)
```
